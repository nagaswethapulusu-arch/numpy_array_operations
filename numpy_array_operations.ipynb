# Task 12 – NumPy Array Operations
Objective: Understand vectorized numerical operations using NumPy.

```python
import numpy as np
```

```python
a = np.array([10, 20, 30, 40, 50])
b = np.array([2, 4, 5, 8, 10])

print("Array a:", a)
print("Array b:", b)
```

```python
print("Shape of a:", a.shape)
print("Shape of b:", b.shape)
print("Same shape:", a.shape == b.shape)
```

```python
print("Addition       (a + b):", a + b)
print("Subtraction    (a - b):", a - b)
print("Multiplication (a * b):", a * b)
print("Division       (a / b):", a / b)
```

```python
print(np.add(a, b))
print(np.subtract(a, b))
print(np.multiply(a, b))
print(np.divide(a, b))
```

```python
print(a * 2)
```

```python
import time

big_a = np.arange(1_000_000)
big_b = np.arange(1_000_000)
list_a, list_b = big_a.tolist(), big_b.tolist()

start = time.perf_counter()
_ = [x + y for x, y in zip(list_a, list_b)]
list_time = time.perf_counter() - start

start = time.perf_counter()
_ = big_a + big_b
numpy_time = time.perf_counter() - start

print(f"List loop : {list_time:.5f} s")
print(f"NumPy     : {numpy_time:.5f} s")
print(f"NumPy is ~{list_time/numpy_time:.0f}x faster")
```

A vectorized operation applies an operation to a whole array at once instead of
a Python for loop. `a + b` adds a[i] + b[i] for every index in compiled C code,
so it is shorter and much faster than looping over Python lists.
